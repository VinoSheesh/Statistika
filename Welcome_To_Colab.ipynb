```
# 1. Import library yang dibutuhkan
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# (Opsional) Mengatur gaya tampilan grafik agar lebih estetik
sns.set_theme(style="whitegrid")

# 2. Asumsikan dataset kamu sudah dimuat ke dalam variabel bernama 'df'
df = pd.read_csv('https://raw.githubusercontent.com/VinoSheesh/Statistika/refs/heads/main/laptop_price%20-%20dataset%20(2).csv?token=GHSAT0AAAAAAEGE6MESUGSADSKVVR2UFI6M2W6VRNQ')

# 3. Membuat Grafik Distribusi (Histogram + Kurva KDE)
plt.figure(figsize=(10, 6))
sns.histplot(df['Price (Euro)'], kde=True, color='blue', bins=30)

# Memberi judul dan label pada grafik
plt.title('Distribusi Harga Laptop (Price in Euro)', fontsize=14, fontweight='bold')
plt.xlabel('Harga (Euro)', fontsize=12)
plt.ylabel('Frekuensi (Jumlah Laptop)', fontsize=12)

# Menampilkan grafik
plt.show()
```

```
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(14, 6))

# Membuat bar plot (secara otomatis menghitung rata-rata / estimator mean)
sns.barplot(x='Company', y='Price (Euro)', data=df, palette='viridis', ci=None)

plt.title('Rata-rata Harga Laptop Berdasarkan Merek (Company)', fontsize=14, fontweight='bold')
plt.xlabel('Merek Laptop (Company)', fontsize=12)
plt.ylabel('Rata-rata Harga (Euro)', fontsize=12)
plt.xticks(rotation=45)

plt.show()
```
